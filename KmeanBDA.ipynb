```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
X = [[1, 2],[2, 1], [2, 3],[8, 8], [9, 8 ], [8, 9],
     [15 ,2], [16, 3], [15, 4]]
```

```python
model = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)
```

```python
model.fit(X)
```

```python
print("Cluster Lables:")
print(model.labels_)
```

```python
print("\nCluster Centroids:")
print(model.cluster_centers_)
```

```python
plt.scatter(
    [p[0] for p in X],
    [p[1] for p in X],
    c=model.labels_
)
plt.scatter(
    model.cluster_centers_[:, 0],
    model.cluster_centers_[:, 1],
    marker="x",
    s=100,
    label="Centroids"
)

plt.title("K-Means Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.show()
```
